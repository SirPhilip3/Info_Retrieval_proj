```python
from index.inverted_index import Index
from index.documents import load_dataset, tokenizer

msmarco = load_dataset("msmarco-passage/dev/judged")
```

```python
# NUM_DOCS = -1

NUM_DOCS = 100
K = 10
```

```python
def check_correctness(original_index, loaded_index, with_terms=False):
    for i, term in enumerate(original_index.index.keys()):
        original_postings = original_index.index[term]
        if with_terms:
            term = str(i)
        loaded_postings = loaded_index.index[term]
        
        if original_postings != loaded_postings:
            print(f"Mismatch for term '{term}':")
            print(f"Original: {original_postings}")
            print(f"Loaded: {loaded_postings}")
            return False
    return True
```

```python
msmarco_index_pfor = Index()
msmarco_index_pfor.clear_index()
msmarco_index_pfor.build_index(msmarco, tokenizer, max_num_docs=NUM_DOCS)
msmarco_index_pfor.remove_tf()
print(msmarco_index_pfor.index.get("the", []))
msmarco_index_pfor.compress_index(method="gamma")
print(msmarco_index_pfor.index.get("the", []))

msmarco_index_pfor.save_index("msmarco_index_pfor.bin")
# msmarco_index_pfor.decompress_index()

reloaded_index_pfor = Index()
reloaded_index_pfor.load_index("msmarco_index_pfor.bin")
if check_correctness(msmarco_index_pfor, reloaded_index_pfor):
    print("The original and loaded indexes match!")
else:
    print("The original and loaded indexes do NOT match!")
```

```python
msmarco_index_pfor = Index()
msmarco_index_pfor.clear_index()
msmarco_index_pfor.build_index(msmarco, tokenizer, max_num_docs=NUM_DOCS)
msmarco_index_pfor.remove_tf()
print(msmarco_index_pfor.index.get("the", []))
msmarco_index_pfor.compress_index(method="pfor_delta", k=4, block_size=32, extra_strategy="None")
print(msmarco_index_pfor.index.get("the", []))

msmarco_index_pfor.save_index("msmarco_index_pfor.bin")

reloaded_index_pfor = Index()
reloaded_index_pfor.load_index("msmarco_index_pfor.bin")

msmarco_index_pfor.decompress_index()
if check_correctness(msmarco_index_pfor, reloaded_index_pfor):
    print("The original and loaded indexes match!")
else:
    print("The original and loaded indexes do NOT match!")

print(msmarco_index_pfor.index)
print(reloaded_index_pfor.index)

print("Original index size:", msmarco_index_pfor.get_approx_size())
print("Loaded index size:", reloaded_index_pfor.get_approx_size())
```
